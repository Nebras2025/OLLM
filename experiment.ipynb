```python
import ollama

# Initialize the streaming generation request
response = ollama.generate(
    model="llama3.1:8b", 
    prompt="whats your name?", 
    stream=True
)

# Iterate over the stream chunks and print correctly
for chunk in response:
    print(chunk['response'], end="", flush=True)

print() # Final newline when generation completes
```

```python
response = ollama.generate(
    model="llama3.1:8b", 
    prompt="whats your name?", 
    options={
        "temperature":0.3,
        "top_p":0.5,
        "top_k":45
    }
)

print(response.response)
```

```python
# Define the conversational history
messages = [
    {"role": "system", "content": "You are a helpful assistant who speaks concisely."},
    {"role": "user", "content": "What is the capital of France?"}
]

# Send the array to the model
response = ollama.chat(model='llama3.1:8b', messages=messages)

# Extract and print the content
print(response['message']['content'])
```
